## Creating the Classification Target

```python
def classify_risk(prob):
    if prob < 0.40:
        return "Low"
    elif prob < 0.70:
        return "Medium"
    else:
        return "High"

df["FloodRisk"] = df["FloodProbability"].apply(classify_risk)

df[["FloodProbability", "FloodRisk"]].head()
```
